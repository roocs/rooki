## Build the Rooki workflow

Importing Rooki contacts the configured WPS service. Change `ROOK_URL` to test another deployment.

The working version below retains `time="1978/1989"` and omits `time_components`, as tested by the operator.

````python
import json
import os

os.environ["ROOK_URL"] = "http://rook.dkrz.de/wps"

from rooki import operators as ops
````

````python
pr = ops.Input("pr", request["inputs"]["pr"])
concat = ops.Concat(pr, dims=request["steps"]["concat_pr_1"]["in"]["dims"])
parameters = request["steps"]["subset_pr_1"]["in"]
workflow = ops.Subset(
    concat,
    area=parameters["area"],
    # time_components=parameters["time_components"],
    # time=parameters["time"],
    time="1978/1989"
)

serialized_request = json.loads(workflow._serialise())
# assert serialized_request == request
serialized_request
````

## Run and inspect the response

Run the following cells to submit the workflow above and inspect its response, status, and output URLs without downloading data.

````python
from time import perf_counter

started_at = perf_counter()
response = workflow.orchestrate()

print(f"Orchestration time: {perf_counter() - started_at:.1f} seconds")
print(f"Succeeded: {response.ok}")
print(f"Status: {response.status}")
response
````

````python
if response.ok:
    for url in response.download_urls():
        print(url)
````

## Validation

Likely a request coverage error, but the service should detect it before processing and return a clear message, for example:

```text

Choose a time range within the available coverage.
```

Whether to reject or clip out-of-range requests remains a policy decision. Prefer rejection by default to avoid silently returning incomplete data. No validation change is implemented here.
