# CMIP6 CanESM5-CanOE tasmin time selection error

When: 2026-08-20

This notebook reproduces the reported error for the following dataset:

`c3s-cmip6.ScenarioMIP.CCCma.CanESM5-CanOE.ssp245.r1i1p2f1.Amon.tasmin.gn.v20190429`

The request selects the years 2015, 2030, 2050, and 2100 within the time range 2015–2100. The reported reason is:

`No timesteps are matching the selection criteria`

The exact `subset` operation is built and submitted below as an orchestrated workflow.

## Configuration

```python
import json
import os

os.environ["ROOK_URL"] = "http://rook.dkrz.de/wps"

from rooki import operators as ops
```

```python
collection = [
    "c3s-cmip6.ScenarioMIP.CCCma.CanESM5-CanOE."
    "ssp245.r1i1p2f1.Amon.tasmin.gn.v20190429"
]
time = "2040/2070"
# time_components = "year:2015,2030,2050,2100"
# time_components = "month:oct"
# time_components = "month:oct|year:2020,2030,2040"
time_components = "year:2040,2060,2070"
```

## Orchestrated subset workflow

Build and inspect the equivalent workflow before submitting it to the orchestrator.

```python
tasmin = ops.Input("tasmin", collection)
workflow = ops.Subset(
    tasmin,
    time=time,
    time_components=time_components,
)

serialized_request = json.loads(workflow._serialise())
serialized_request
```

```python
resp = workflow.orchestrate()
resp.ok, resp.status
```
