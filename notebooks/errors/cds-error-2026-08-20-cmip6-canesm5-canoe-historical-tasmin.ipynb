# CMIP6 CanESM5-CanOE historical tasmin year-selection comparison

When: 2026-08-20

This notebook provides a successful comparison for the ScenarioMIP `tasmin` year-selection error. It uses the corresponding historical CanESM5-CanOE monthly dataset:

`c3s-cmip6.CMIP.CCCma.CanESM5-CanOE.historical.r1i1p2f1.Amon.tasmin.gn.v20190429`

The request selects the years 1980, 1990, 2000, and 2010 within the time range 1980–2010. The reported result is one successful request:

`requests=1 outcomes={'successful': 1}`

Only the orchestrated `subset` workflow is used so its behavior can be compared directly with the failing ScenarioMIP notebook.

## Configuration

```python
import json
import os

os.environ["ROOK_URL"] = "http://rook.dkrz.de/wps"

from rooki import operators as ops
```

```python
collection = [
    "c3s-cmip6.CMIP.CCCma.CanESM5-CanOE."
    "historical.r1i1p2f1.Amon.tasmin.gn.v20190429"
]
time = "1980/2010"
time_components = "year:1980,1990,2000,2010"
```

## Orchestrated subset workflow

Build and inspect the workflow before submitting it to the orchestrator.

```python
tasmin = ops.Input("tasmin", collection)
workflow = ops.Subset(
    tasmin,
    time=time,
    time_components=time_components,
)

serialized_request = json.loads(workflow._serialise())
serialized_request
```

```python
resp = workflow.orchestrate()
resp.ok, resp.status
```
